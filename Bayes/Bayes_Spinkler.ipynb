```python
from pgmpy.models import BayesianNetwork
from pgmpy.factors.discrete import TabularCPD
```

```python
# Defining the model structure. We can define the network by just passing a list of edges.
model = BayesianNetwork([('C', 'S'), ('C', 'R'), ('S', 'W'), ('R', 'W')])
print(model.nodes())
print(model.edges())
```

```python
# Defining individual Conditional Probability Distribution (CPD).
cpd_c = TabularCPD(variable='C', variable_card=2, values=[[0.5], [0.5]])
print(cpd_c)
```

```python
cpd_s = TabularCPD(variable='S', variable_card=2, 
                   values=[[0.5, 0.9],
                           [0.5, 0.1]],
                  evidence=['C'],
                  evidence_card=[2])
print(cpd_s)
```

```python
cpd_r = TabularCPD(variable='R', variable_card=2, 
                   values=[[0.8, 0.2],
                           [0.2, 0.8]],
                  evidence=['C'],
                  evidence_card=[2])

cpd_w = TabularCPD(variable='W', variable_card=2, 
                   values=[[1.0, 0.1, 0.1, 0.01],
                           [0.0, 0.9, 0.9, 0.99]],
                  evidence=['S', 'R'],
                  evidence_card=[2, 2])
```

```python
# Create the Bayesian network
model.add_cpds(cpd_c,cpd_s,cpd_r,cpd_w)
# model.get_cpds
model.check_model()
# model.get_independencies()
```

```python
# Inference
from pgmpy.inference import VariableElimination
infer = VariableElimination(model)
w_dist = infer.query(['W'])
print(w_dist)
```

```python
w2_dist = infer.query(['W'], evidence={'C': 0,'R': 0})
print(w2_dist)
```
